```python
# ============================================================
# FOOD DELIVERY DATA ANALYSIS PROJECT
# Python Analysis / EDA
# ============================================================

import pandas as pd
import numpy as np
from google.colab import files

# ============================================================
# 1. LOAD CLEANED DATASET
# ============================================================

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

display(df.head())


# ============================================================
# 2. DATASET STRUCTURE
# ============================================================

print("\n" + "=" * 60)
print("DATASET STRUCTURE")
print("=" * 60)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Shape:")
print(df.shape)


# ============================================================
# 3. BASIC DATA INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("BASIC DATA INFORMATION")
print("=" * 60)

df.info()


# ============================================================
# 4. MISSING VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUE ANALYSIS")
print("=" * 60)

missing = df.isnull().sum()

missing_percentage = (
    df.isnull().mean() * 100
).round(2)

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Values": missing.values,
    "Missing_Percentage": missing_percentage.values
})

display(missing_report)


# ============================================================
# 5. DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)


# ============================================================
# 6. UNIQUE VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("UNIQUE VALUE ANALYSIS")
print("=" * 60)

unique_report = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": [df[col].nunique() for col in df.columns]
})

display(unique_report)


# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("DESCRIPTIVE STATISTICS")
print("=" * 60)

numeric_df = df.select_dtypes(include=np.number)

display(numeric_df.describe().T)


# ============================================================
# 8. MEDIAN, MODE, VARIANCE & STANDARD DEVIATION
# ============================================================

print("\n" + "=" * 60)
print("STATISTICAL ANALYSIS")
print("=" * 60)

statistical_analysis = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Mode": numeric_df.mode().iloc[0],
    "Standard_Deviation": numeric_df.std(),
    "Variance": numeric_df.var(),
    "Minimum": numeric_df.min(),
    "Maximum": numeric_df.max()
})

display(statistical_analysis)


# ============================================================
# 9. SKEWNESS
# ============================================================

print("\n" + "=" * 60)
print("SKEWNESS ANALYSIS")
print("=" * 60)

skewness = numeric_df.skew()

skewness_report = pd.DataFrame({
    "Column": skewness.index,
    "Skewness": skewness.values
})

display(skewness_report)


# ============================================================
# 10. KURTOSIS
# ============================================================

print("\n" + "=" * 60)
print("KURTOSIS ANALYSIS")
print("=" * 60)

kurtosis = numeric_df.kurtosis()

kurtosis_report = pd.DataFrame({
    "Column": kurtosis.index,
    "Kurtosis": kurtosis.values
})

display(kurtosis_report)


# ============================================================
# 11. ORDER STATUS ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("ORDER STATUS ANALYSIS")
print("=" * 60)

status_analysis = (
    df.groupby("order_status")
    .agg(
        Total_Orders=("order_id", "count"),
        Total_Order_Value=("order_amount", "sum"),
        Average_Order_Value=("order_amount", "mean")
    )
    .reset_index()
    .sort_values("Total_Orders", ascending=False)
)

display(status_analysis)


# ============================================================
# 12. CITY-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CITY-WISE ANALYSIS")
print("=" * 60)

city_analysis = (
    df.groupby("city")
    .agg(
        Total_Orders=("order_id", "count"),
        Total_Sales=("order_amount", "sum"),
        Average_Order_Value=("order_amount", "mean"),
        Average_Delivery_Time=("delivery_time_min", "mean"),
        Average_Rating=("restaurant_rating", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

display(city_analysis)


# ============================================================
# 13. CUISINE-WISE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CUISINE-WISE ANALYSIS")
print("=" * 60)

cuisine_analysis = (
    df.groupby("cuisine")
    .agg(
        Total_Orders=("order_id", "count"),
        Total_Sales=("order_amount", "sum"),
        Average_Order_Value=("order_amount", "mean"),
        Average_Rating=("restaurant_rating", "mean")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=False)
)

display(cuisine_analysis)


# ============================================================
# 14. PAYMENT METHOD ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("PAYMENT METHOD ANALYSIS")
print("=" * 60)

payment_analysis = (
    df.groupby("payment_method")
    .agg(
        Total_Orders=("order_id", "count"),
        Total_Sales=("order_amount", "sum"),
        Average_Order_Value=("order_amount", "mean")
    )
    .reset_index()
    .sort_values("Total_Orders", ascending=False)
)

display(payment_analysis)


# ============================================================
# 15. DELIVERY PERFORMANCE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DELIVERY PERFORMANCE ANALYSIS")
print("=" * 60)

delivery_analysis = pd.DataFrame({
    "Metric": [
        "Average Delivery Time",
        "Median Delivery Time",
        "Minimum Delivery Time",
        "Maximum Delivery Time",
        "Average Delivery Distance",
        "Maximum Delivery Distance"
    ],
    "Value": [
        df["delivery_time_min"].mean(),
        df["delivery_time_min"].median(),
        df["delivery_time_min"].min(),
        df["delivery_time_min"].max(),
        df["delivery_distance_km"].mean(),
        df["delivery_distance_km"].max()
    ]
})

display(delivery_analysis)


# ============================================================
# 16. CUSTOMER / ORDER VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("ORDER VALUE ANALYSIS")
print("=" * 60)

order_value_analysis = pd.DataFrame({
    "Metric": [
        "Total Order Amount",
        "Average Order Amount",
        "Median Order Amount",
        "Minimum Order Amount",
        "Maximum Order Amount",
        "Total Delivery Fee",
        "Total Discount"
    ],
    "Value": [
        df["order_amount"].sum(),
        df["order_amount"].mean(),
        df["order_amount"].median(),
        df["order_amount"].min(),
        df["order_amount"].max(),
        df["delivery_fee"].sum(),
        df["discount"].sum()
    ]
})

display(order_value_analysis)


# ============================================================
# 17. NET ORDER VALUE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("NET ORDER VALUE ANALYSIS")
print("=" * 60)

# Recalculate to ensure consistency
df["net_order_value"] = (
    df["order_amount"]
    + df["delivery_fee"]
    - df["discount"]
)

print("Total Net Order Value:",
      round(df["net_order_value"].sum(), 2))

print("Average Net Order Value:",
      round(df["net_order_value"].mean(), 2))

print("Median Net Order Value:",
      round(df["net_order_value"].median(), 2))


# ============================================================
# 18. DISCOUNT ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DISCOUNT ANALYSIS")
print("=" * 60)

discount_analysis = pd.DataFrame({
    "Metric": [
        "Total Discount",
        "Average Discount",
        "Median Discount",
        "Maximum Discount"
    ],
    "Value": [
        df["discount"].sum(),
        df["discount"].mean(),
        df["discount"].median(),
        df["discount"].max()
    ]
})

display(discount_analysis)


# ============================================================
# 19. DELIVERY FEE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("DELIVERY FEE ANALYSIS")
print("=" * 60)

delivery_fee_analysis = pd.DataFrame({
    "Metric": [
        "Total Delivery Fee",
        "Average Delivery Fee",
        "Median Delivery Fee",
        "Minimum Delivery Fee",
        "Maximum Delivery Fee"
    ],
    "Value": [
        df["delivery_fee"].sum(),
        df["delivery_fee"].mean(),
        df["delivery_fee"].median(),
        df["delivery_fee"].min(),
        df["delivery_fee"].max()
    ]
})

display(delivery_fee_analysis)


# ============================================================
# 20. RATING ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("RESTAURANT RATING ANALYSIS")
print("=" * 60)

rating_analysis = pd.DataFrame({
    "Metric": [
        "Average Rating",
        "Median Rating",
        "Minimum Rating",
        "Maximum Rating"
    ],
    "Value": [
        df["restaurant_rating"].mean(),
        df["restaurant_rating"].median(),
        df["restaurant_rating"].min(),
        df["restaurant_rating"].max()
    ]
})

display(rating_analysis)


# ============================================================
# 21. MONTH-WISE ANALYSIS
# ============================================================

df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()

print("\n" + "=" * 60)
print("MONTH-WISE ANALYSIS")
print("=" * 60)

monthly_analysis = (
    df.groupby(["year", "month", "month_name"])
    .agg(
        Total_Orders=("order_id", "count"),
        Total_Sales=("order_amount", "sum"),
        Average_Order_Value=("order_amount", "mean")
    )
    .reset_index()
    .sort_values(["year", "month"])
)

display(monthly_analysis)


# ============================================================
# 22. TOP CITIES BY SALES
# ============================================================

print("\n" + "=" * 60)
print("TOP 5 CITIES BY SALES")
print("=" * 60)

top_cities = (
    df.groupby("city")["order_amount"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

display(top_cities)


# ============================================================
# 23. TOP CUISINES BY SALES
# ============================================================

print("\n" + "=" * 60)
print("TOP CUISINES BY SALES")
print("=" * 60)

top_cuisines = (
    df.groupby("cuisine")["order_amount"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

display(top_cuisines)


# ============================================================
# 24. TOP ORDERS BY VALUE
# ============================================================

print("\n" + "=" * 60)
print("TOP 10 ORDERS BY ORDER VALUE")
print("=" * 60)

top_orders = (
    df[
        [
            "order_id",
            "city",
            "cuisine",
            "order_amount",
            "delivery_fee",
            "discount",
            "net_order_value"
        ]
    ]
    .sort_values("order_amount", ascending=False)
    .head(10)
)

display(top_orders)


# ============================================================
# 25. CORRELATION ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

correlation = numeric_df.corr()

display(correlation)


# ============================================================
# 26. STRONG CORRELATIONS
# ============================================================

print("\n" + "=" * 60)
print("IMPORTANT CORRELATIONS")
print("=" * 60)

corr_pairs = (
    correlation
    .where(np.triu(np.ones(correlation.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

corr_pairs.columns = ["Variable_1", "Variable_2", "Correlation"]

corr_pairs["Absolute_Correlation"] = (
    corr_pairs["Correlation"].abs()
)

corr_pairs = corr_pairs.sort_values(
    "Absolute_Correlation",
    ascending=False
)

display(corr_pairs.head(10))


# ============================================================
# 27. OUTLIER ANALYSIS USING IQR
# ============================================================

print("\n" + "=" * 60)
print("OUTLIER ANALYSIS")
print("=" * 60)

outlier_results = []

for col in numeric_df.columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)


# ============================================================
# 28. HIGH VALUE ORDERS
# ============================================================

print("\n" + "=" * 60)
print("HIGH VALUE ORDER ANALYSIS")
print("=" * 60)

high_value_threshold = df["order_amount"].quantile(0.90)

high_value_orders = df[
    df["order_amount"] >= high_value_threshold
]

print(
    "High Value Order Threshold:",
    round(high_value_threshold, 2)
)

print(
    "Number of High Value Orders:",
    len(high_value_orders)
)

print(
    "Percentage of High Value Orders:",
    round(
        len(high_value_orders) / len(df) * 100,
        2
    ),
    "%"
)


# ============================================================
# 29. DELIVERY TIME CATEGORIES
# ============================================================

def delivery_category(time):

    if time <= 30:
        return "Fast"

    elif time <= 45:
        return "Moderate"

    else:
        return "Slow"


df["delivery_category"] = (
    df["delivery_time_min"]
    .apply(delivery_category)
)

delivery_category_analysis = (
    df.groupby("delivery_category")
    .agg(
        Total_Orders=("order_id", "count"),
        Average_Order_Value=("order_amount", "mean"),
        Average_Delivery_Time=("delivery_time_min", "mean")
    )
    .reset_index()
)

print("\n" + "=" * 60)
print("DELIVERY CATEGORY ANALYSIS")
print("=" * 60)

display(delivery_category_analysis)


# ============================================================
# 30. FINAL BUSINESS KPI SUMMARY
# ============================================================

total_orders = len(df)

total_sales = df["order_amount"].sum()

total_delivery_fee = df["delivery_fee"].sum()

total_discount = df["discount"].sum()

total_net_value = df["net_order_value"].sum()

average_order_value = df["order_amount"].mean()

average_delivery_time = df["delivery_time_min"].mean()

average_rating = df["restaurant_rating"].mean()

print("\n")
print("=" * 60)
print("FINAL BUSINESS KPI SUMMARY")
print("=" * 60)

print(f"Total Orders          : {total_orders}")
print(f"Total Sales           : {total_sales:,.2f}")
print(f"Total Delivery Fee    : {total_delivery_fee:,.2f}")
print(f"Total Discount        : {total_discount:,.2f}")
print(f"Total Net Order Value : {total_net_value:,.2f}")
print(f"Average Order Value   : {average_order_value:,.2f}")
print(f"Average Delivery Time : {average_delivery_time:,.2f} minutes")
print(f"Average Rating        : {average_rating:,.2f}")


# ============================================================
# 31. EXPORT ANALYSIS RESULTS TO EXCEL
# ============================================================

analysis_file = "Python_Analysis_Results.xlsx"

with pd.ExcelWriter(analysis_file, engine="openpyxl") as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned_Data",
        index=False
    )

    statistical_analysis.to_excel(
        writer,
        sheet_name="Statistics"
    )

    city_analysis.to_excel(
        writer,
        sheet_name="City_Analysis",
        index=False
    )

    cuisine_analysis.to_excel(
        writer,
        sheet_name="Cuisine_Analysis",
        index=False
    )

    payment_analysis.to_excel(
        writer,
        sheet_name="Payment_Analysis",
        index=False
    )

    status_analysis.to_excel(
        writer,
        sheet_name="Status_Analysis",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly_Analysis",
        index=False
    )

    outlier_report.to_excel(
        writer,
        sheet_name="Outlier_Analysis",
        index=False
    )

    correlation.to_excel(
        writer,
        sheet_name="Correlation"
    )

    delivery_category_analysis.to_excel(
        writer,
        sheet_name="Delivery_Performance",
        index=False
    )

print("\n" + "=" * 60)
print("PYTHON ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nAnalysis report saved as:")
print(analysis_file)

files.download(analysis_file)
```
